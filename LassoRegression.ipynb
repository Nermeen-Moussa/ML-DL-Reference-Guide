**performs regularization and automatic feature selection by adding an L1 penalty to the loss function**

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
ca_housing = fetch_california_housing()
X=ca_housing.data
y=ca_housing.target
ca_housing.target_names
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=19)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
lasso=Lasso()
lasso.fit(X_train,y_train)
y_pred=lasso.predict(X_test)
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```

```python
# Regularization Penalty
param_grid={
    'alpha':[0.0001,0.001,0.01,0.1,1,10,100]
}
lasso_cv=GridSearchCV(lasso,param_grid,cv=5)
lasso_cv.fit(X_train,y_train)
print(lasso_cv.best_estimator_)
print(lasso_cv.best_score_)
```

```python
lasso3=Lasso(alpha=0.001)
lasso3.fit(X_train,y_train)
y_pred=lasso3.predict(X_test)
lasso3.intercept_
lasso3.coef_
```

```python
feature_names=list(ca_housing.feature_names)
df=pd.DataFrame(lasso3.coef_,index=feature_names,columns=['Coefficient'])
df.sort_values(by='Coefficient',ascending=False)
```
