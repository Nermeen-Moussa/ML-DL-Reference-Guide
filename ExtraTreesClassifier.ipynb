**fits multiple randomized decision trees on data subsets and uses majority voting for classification**

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.ensemble import ExtraTreesClassifier
```

```python
X,y=make_classification(n_features=11,random_state=21)
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=21)
```

```python
etc=ExtraTreesClassifier(random_state=21)
etc.fit(X_train,y_train)
etc.score(X_test,y_test)
```

```python
cross_val_score(etc,X_train,y_train,scoring='accuracy',cv=5,n_jobs=-1).mean()
```

```python
param_grid={
    'n_estimators':[100,250,500],
    'max_features':[3,5,7,9],
    'min_samples_leaf':[5,15,25],
    'criterion':['gini','entropy']
}
```

```python
etc2=GridSearchCV(etc,param_grid,scoring='accuracy',cv=5,n_jobs=-1)
etc2.fit(X_train,y_train)
etc2.score(X_test,y_test)
etc2.best_params_
```

```python
etc3=ExtraTreesClassifier(n_estimators=250,max_features=3,min_samples_leaf=25,criterion='gini',random_state=21)
etc3.fit(X_train,y_train)
etc3.score(X_test,y_test)
```
